# Building the full labelled dataset

The pipeline has been proven on five files. This notebook scales it to all 66 labelled soundscapes, giving the complete set of 1,478 segments across all nine sites. That is the data the evaluation harness needs, since holding sites out of a five-file sample would leave too little to measure anything.

```python
# Import necessary libraries
import os
import json
import numpy as np
import pandas as pd
import librosa

SR = 32000
DATA_DIR = '../data'

# Load the labelled soundscape segments CSV file and get unique filenames
ss = pd.read_csv('../data/train_soundscapes_labels.csv')
all_files = ss['filename'].unique()

print("labelled soundscape files:", len(all_files))
print("total segments:", len(ss))
```

## Processing every segment

The same cut-and-transform as before, with one improvement. Previously each five second slice was read from disk separately, meaning a file was opened 24 times over. Here each file is loaded once into memory and then sliced, which is considerably faster across 1,478 segments.

The site code is also extracted from each filename and kept alongside the data, since the evaluation split depends on it.

```python
import re

# Function to convert time strings (HH:MM:SS) to seconds
def time_to_seconds(t):
    h, m, s = t.split(':')
    return int(h) * 3600 + int(m) * 60 + int(s)

# Function to create a mel spectrogram from an audio signal
def make_spectrogram(y):
    mel = librosa.feature.melspectrogram(y=y, sr=SR, n_mels=128, fmin=40, fmax=15000)
    return librosa.power_to_db(mel, ref=np.max)

specs, labels, sites, srcfiles = [], [], [], []
# Process each soundscape file and extract segments based on the CSV labels
for n, fname in enumerate(on_disk, 1):
    path = os.path.join(DATA_DIR, fname)
    y_full, _ = librosa.load(path, sr=SR)          # load the whole file once

    site = re.search(r'_(S\d+)_', fname).group(1)  # e.g. S22
    rows = ss[ss['filename'] == fname]

    # Process each segment in the CSV for the current soundscape file
    for _, row in rows.iterrows():
        start = time_to_seconds(row['start'])
        end = time_to_seconds(row['end'])

        y_seg = y_full[start * SR : end * SR]      # slice the array, no re-reading

        target = SR * (end - start)
        if len(y_seg) < target:
            y_seg = np.pad(y_seg, (0, target - len(y_seg)))

        specs.append(make_spectrogram(y_seg))
        labels.append(row['primary_label'].split(';'))
        sites.append(site)
        srcfiles.append(fname)

    print(f"[{n}/{len(on_disk)}] {fname}  ->  {len(rows)} segments")

# Convert the list of spectrograms to a NumPy array and print the shapes of the spectrogram stack and labels
specs = np.array(specs)
print()
print("spectrogram stack:", specs.shape)
print("segments:", len(labels))
```

## Segments per site

The evaluation split holds out whole sites, so it matters how the segments are distributed across them. A site with very few segments would make a poor test set, and a site holding a large share of the data would leave too little for training.

```python
# Count the number of segments per site and print the counts along with the total
site_counts = pd.Series(sites).value_counts().sort_index()
print(site_counts)
print()
print("total:", site_counts.sum())
```

```python
# Save the spectrograms and metadata to files
np.save('../data/full_specs.npy', specs)
with open('../data/full_meta.json', 'w') as f:
    json.dump({'labels': labels, 'sites': sites, 'files': srcfiles}, f)
print("saved")
```
